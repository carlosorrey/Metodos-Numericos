# Método de Bisección

Implementación del Algoritmo 2.1 de Burden usando el Ejemplo 1.

## Problema

Encontrar una raíz de:

$$f(x)=x^3+4x^2-10$$

usando el intervalo $[1,2]$.

Como $f(1)=-5$ y $f(2)=14$, hay un cambio de signo y podemos aplicar bisección.

## Método

El intervalo se divide a la mitad. El punto medio es

$$p=a+\frac{b-a}{2}$$

Después se conserva la mitad donde sigue existiendo el cambio de signo. Usaremos $TOL=10^{-4}$.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3 + 4*x**2 - 10
```

## Algoritmo de bisección

La función siguiente sigue los pasos del Algoritmo 2.1 y guarda los resultados de las iteraciones.

```python
def biseccion(f, a, b, tol, N0):
    fa = f(a)
    datos = []

    if fa * f(b) >= 0:
        print('No hay cambio de signo en el intervalo')
        return None, datos

    for i in range(1, N0 + 1):
        p = a + (b - a) / 2
        fp = f(p)
        error = (b - a) / 2
        datos.append([i, a, b, p, fp, error])

        if fp == 0 or error < tol:
            return p, datos

        if fa * fp > 0:
            a = p
            fa = fp
        else:
            b = p

    return p, datos
```

## Ejemplo 1

Usamos $a=1$, $b=2$, $TOL=10^{-4}$ y $N_0=100$.

```python
a = 1
b = 2
tol = 1e-4
N0 = 100

solucion, datos = biseccion(f, a, b, tol, N0)

print('Solución aproximada:', solucion)
print('Número de iteraciones:', len(datos))
```

## Resultados

Tabla de las iteraciones.

```python
tabla = pd.DataFrame(datos, columns=['i', 'a', 'b', 'p', 'f(p)', '(b-a)/2'])
tabla
```

## Gráfica

Se muestran los puntos medios de cada iteración.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[3] for fila in datos]

plt.plot(iteraciones, puntos, 'o-')
plt.xlabel('Iteración')
plt.ylabel('p')
plt.title('Método de bisección')
plt.grid()
plt.show()
```

## Comprobación

Evaluamos la función en la aproximación encontrada.

```python
print('p =', solucion)
print('f(p) =', f(solucion))
```

## Conclusión

Con una tolerancia de $10^{-4}$ se obtiene la aproximación

$$\boxed{p\approx1.3651733398}$$

El valor de $f(p)$ es cercano a cero, por lo que la aproximación es correcta para la tolerancia utilizada.
