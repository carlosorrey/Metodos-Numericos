
# Diferencias divididas de Newton

**Algoritmo 3.2 de Burden — Ejemplo 1**

En este trabajo se calcula el polinomio de interpolación de Newton con los datos del ejemplo 1 de la página 94.



## 1. Fórmula

El algoritmo calcula las diferencias divididas con la fórmula:

$$
F_{i,j}=\frac{F_{i,j-1}-F_{i-1,j-1}}{x_i-x_{i-j}}
$$

Con los coeficientes obtenidos se forma el polinomio:

$$
P_n(x)=F_{0,0}+\sum_{i=1}^{n}F_{i,i}\prod_{j=0}^{i-1}(x-x_j)
$$



## 2. Importar la librería

Usaremos `sympy` para escribir el polinomio y hacer su gráfica.


```python
import sympy as sp
```


## 3. Datos del ejemplo

Los valores se toman de la tabla 3.10 de Burden.


```python

x = [1.0, 1.3, 1.6, 1.9, 2.2]
y = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]
```


## 4. Calcular las diferencias divididas

La función guarda los cálculos en una tabla. En cada columna se obtiene un orden más de diferencias divididas.


```python

def diferencias_divididas(x, y):
    n = len(x)
    F = [[0.0 for j in range(n)] for i in range(n)]

    for i in range(n):
        F[i][0] = y[i]

    for j in range(1, n):
        for i in range(j, n):
            F[i][j] = (F[i][j-1] - F[i-1][j-1]) / (x[i] - x[i-j])

    coeficientes = [F[i][i] for i in range(n)]
    return F, coeficientes
```


## 5. Mostrar la tabla y los coeficientes

Se imprime la tabla para revisar los resultados. Los valores de la diagonal son los coeficientes que se usan en el polinomio de Newton.


```python

tabla, coeficientes = diferencias_divididas(x, y)

print("Tabla de diferencias divididas:")
for i in range(len(x)):
    print([round(tabla[i][j], 7) for j in range(i + 1)])

print("\nCoeficientes:")
for c in coeficientes:
    print(round(c, 7))
```


## 6. Formar el polinomio y graficarlo

Esta función toma los puntos y los coeficientes calculados. Con `sympy` construye el polinomio y después muestra su gráfica junto con los puntos originales.


```python

def polinomio_newton(x, y, coeficientes):
    t = sp.symbols('x')
    P = sp.Float(coeficientes[0])

    for i in range(1, len(coeficientes)):
        termino = sp.Float(coeficientes[i])
        for j in range(i):
            termino = termino * (t - sp.Float(x[j]))
        P = P + termino

    print("Polinomio de Newton:")
    sp.pprint(sp.expand(P))

    sp.plot(P, (t, min(x), max(x)), title="Polinomio de interpolación de Newton")

    return P
```


## 7. Resultado


```python
P = polinomio_newton(x, y, coeficientes)
```


El polinomio obtenido interpola los cinco datos de la tabla 3.10.
