# Método de Bisección

### Ejemplo 1 - Burden

## Problema

En el ejemplo 1 del libro se quiere encontrar una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Comprobamos primero los extremos:

$$f(1)=-5$$
$$f(2)=14$$

Como tienen signos diferentes, podemos usar bisección.

## Método

La bisección consiste en dividir el intervalo en dos partes. Para esto calculamos el punto medio

$$m=a+\frac{b-a}{2}$$

Después vemos si la raíz queda entre $a$ y $m$ o entre $m$ y $b$ y seguimos trabajando con esa mitad.

En el código se usa `error = abs(b-a)` como en el código que vimos en clase.

## Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Función

```python
def funcion1(x):
    return x**3 + 4*x**2 - 10
```

## Código de bisección

Esta parte sigue la idea del código realizado en clase. `fa`, `fb` y `fm` son los valores de la función en los puntos $a$, $b$ y $m$.

```python
def biseccion(a0, b0, tol, nmax):
    # Paso 1: inicialización
    a = a0
    b = b0
    i = 1
    fa = funcion1(a)

    datos = []

    # Paso 2: repetir mientras i <= N0
    while i <= nmax:

        # Paso 3: calcular el punto medio p
        p = a + (b - a)/2
        fp = funcion1(p)

        # En el ejemplo 1 de la página 38 se usa una cota
        # de error relativo para detener el procedimiento.
        error_relativo = ((b - a)/2) / min(abs(a), abs(b))

        datos.append([i, a, fa, b, funcion1(b), p, fp, error_relativo])

        # Paso 4: condición de parada
        if fp == 0 or error_relativo < tol:
            return p, datos

        # Paso 5: aumentar el contador
        i = i + 1

        # Paso 6: conservar el intervalo donde hay cambio de signo
        if fa * fp > 0:
            a = p
            fa = fp
        else:
            b = p

    # Paso 7: si se alcanza el número máximo de iteraciones
    print("El método fracasó después de", nmax, "iteraciones.")
    return p, datos
```

## Aplicación al ejemplo 1

Tomamos el intervalo que aparece en Burden, $[1,2]$. La tolerancia será $10^{-4}$.

```python
a0 = 1
b0 = 2
tol = 10**(-4)
nmax = 100

raiz, datos = biseccion(a0, b0, tol, nmax)

print("La raíz aproximada es:", raiz)
print("Iteraciones:", len(datos))
```

## Resultados de las iteraciones

```python
tabla = pd.DataFrame(datos, columns=[
    'iter', 'a', 'f(a)', 'b', 'f(b)', 'm', 'f(m)', 'error'
])

tabla
```

## Gráfica

Aquí se puede observar cómo se van acercando los puntos medios.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

plt.plot(iteraciones, puntos, 'o-')
plt.xlabel('Iteración')
plt.ylabel('m')
plt.title('Convergencia del método de bisección')
plt.grid()
plt.show()
```

## Comprobación

Evaluamos la función en la aproximación que obtuvimos.

```python
print("m =", raiz)
print("f(m) =", funcion1(raiz))
```

## Resultado final

Para el ejemplo 1 de Burden, utilizando el intervalo $[1,2]$ y la tolerancia de $10^{-4}$, el método de bisección proporciona después de 13 iteraciones:

$$\boxed{p_{13}=1.3651123047}$$

Este resultado coincide con la aproximación indicada en la página 38 del libro.

Además, se tiene:

$$f(p_{13})\approx -0.001943659$$

La aproximación obtenida corresponde a la raíz de

$$f(x)=x^3+4x^2-10.$$
