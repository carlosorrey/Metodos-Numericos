# Método de Bisección

### Ejemplo 1 - Burden

## Problema

En el ejemplo 1 del libro se quiere encontrar una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Comprobamos primero los extremos:

$$f(1)=-5$$
$$f(2)=14$$

Como tienen signos diferentes, podemos usar bisección.

## Método

La bisección consiste en dividir el intervalo en dos partes. Para esto calculamos el punto medio

$$m=a+\frac{b-a}{2}$$

Después vemos si la raíz queda entre $a$ y $m$ o entre $m$ y $b$ y seguimos trabajando con esa mitad.

El código sigue el **Algoritmo 2.1 de Burden**: se calcula el punto medio $p$, se evalúa $f(p)$ y se conserva el subintervalo donde los extremos tienen signos opuestos.


## Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Función

```python
def funcion1(x):
    return x**3 + 4*x**2 - 10
```

## Código de bisección

Esta parte sigue la idea del código realizado en clase. `fa`, `fb` y `fm` son los valores de la función en los puntos $a$, $b$ y $m$.

```python
def biseccion(a0, b0, tol, nmax):
    # Paso 1: valores iniciales del Algoritmo 2.1
    a = a0
    b = b0
    i = 1
    FA = funcion1(a)
    datos = []

    # Paso 2: repetir mientras i <= N0
    while i <= nmax:
        # Paso 3: calcular el punto medio p
        p = a + (b - a)/2
        FP = funcion1(p)

        # Guardamos los valores de la iteración
        datos.append([i, a, FA, b, funcion1(b), p, FP, abs(b-a)])

        # Si f(p) = 0, se encontró la raíz
        if FP == 0:
            return p, datos

        # Paso 5: aumentar el contador
        i = i + 1

        # Paso 6: conservar el intervalo que contiene la raíz
        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

        # Para el ejemplo 1 de la página 38,
        # Burden usa esta cota de error relativo.
        cota_error = abs(b-a) / min(abs(a), abs(b))
        if cota_error < tol:
            return p, datos

    # Paso 7: no se alcanzó la tolerancia dentro de N0 iteraciones
    return p, datos
```

## Aplicación al ejemplo 1

Tomamos el intervalo que aparece en Burden, $[1,2]$. La tolerancia será $10^{-4}$.

```python
a0 = 1
b0 = 2
tol = 10**(-4)
nmax = 100

raiz, datos = biseccion(a0, b0, tol, nmax)

print("La raíz aproximada es:", raiz)
print("Iteraciones:", len(datos))
```

## Resultados de las iteraciones

```python
tabla = pd.DataFrame(datos, columns=[
    'iter', 'a', 'f(a)', 'b', 'f(b)', 'm', 'f(m)', 'error'
])

tabla
```

## Gráfica

Aquí se puede observar cómo se van acercando los puntos medios.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

plt.plot(iteraciones, puntos, 'o-')
plt.xlabel('Iteración')
plt.ylabel('m')
plt.title('Convergencia del método de bisección')
plt.grid()
plt.show()
```

## Comprobación

Evaluamos la función en la aproximación que obtuvimos.

```python
print("m =", raiz)
print("f(m) =", funcion1(raiz))
```

## Resultado final

Para el ejemplo 1, Burden utiliza como criterio de parada la cota del **error relativo** indicada en la página 38:

$$\frac{|p-p_n|}{\min\{|a_n|,|b_n|\}}<10^{-4}.$$

Después de 13 iteraciones se obtiene

$$\boxed{p_{13}=1.3651123047}$$

Por lo tanto, esta es la aproximación de la raíz solicitada en el ejemplo 1.
