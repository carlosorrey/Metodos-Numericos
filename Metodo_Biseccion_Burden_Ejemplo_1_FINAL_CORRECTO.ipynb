# Método de Bisección

### Ejemplo 1 - Burden

## Problema

En el ejemplo 1 del libro se quiere encontrar una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Comprobamos primero los extremos:

$$f(1)=-5$$
$$f(2)=14$$

Como tienen signos diferentes, podemos usar bisección.

## Método

La bisección consiste en dividir el intervalo en dos partes. Para esto calculamos el punto medio

$$m=a+\frac{b-a}{2}$$

Después vemos si la raíz queda entre $a$ y $m$ o entre $m$ y $b$ y seguimos trabajando con esa mitad.

El código sigue el **Algoritmo 2.1 de Burden**: se calcula el punto medio $p$, se evalúa $f(p)$ y se conserva el subintervalo donde los extremos tienen signos opuestos.


## Librerías

Se utilizan las mismas librerías empleadas en el código visto en clase: `math`, `numpy` y `matplotlib`.

```python
from math import *
import numpy as np
import matplotlib.pyplot as PLT
```

## Función

```python
def funcion1(x):
    return x**3 + 4*x**2 - 10
```

## Código de bisección

El código conserva la estructura del método de bisección trabajado en clase. Para la prueba con Burden se cambia únicamente la función y el intervalo, y se utiliza el criterio de error relativo indicado en la página 38 para reproducir el resultado del Ejemplo 1.

```python
def biseccion(a0, b0, tol, nmax):
    # Paso 1 del Algoritmo 2.1
    a = a0
    b = b0
    i = 1
    FA = funcion1(a)
    datos = []

    # Paso 2: mientras i <= N0 se realizan los pasos 3-6
    while i <= nmax:
        # Paso 3: calcular p y FP
        p = a + (b - a)/2
        FP = funcion1(p)

        datos.append([i, a, FA, b, funcion1(b), p, FP, (b-a)/2])

        # Paso 4: criterio de parada del Algoritmo 2.1
        if FP == 0 or (b - a)/2 < tol:
            return p, datos

        # Paso 5
        i = i + 1

        # Paso 6: conservar el intervalo con cambio de signo
        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

    # Paso 7: se alcanzó el máximo de iteraciones
    return p, datos
```

### Prueba con el criterio de parada del Ejemplo 1

En la página 38, Burden indica que en este ejemplo la iteración termina cuando la cota del error relativo es menor que $10^{-4}$:

$$\frac{|p-p_n|}{\min\{|a_n|,|b_n|\}}<10^{-4}.$$

Por eso, para reproducir exactamente el resultado de la tabla 2.1, se conserva el mismo procedimiento de bisección y se cambia únicamente el criterio de parada, tal como permite el texto después del Algoritmo 2.1.

```python
def biseccion_ejemplo1(a0, b0, tol, nmax):
    a = a0
    b = b0
    i = 1
    FA = funcion1(a)
    datos = []

    while i <= nmax:
        p = a + (b - a)/2
        FP = funcion1(p)

        datos.append([i, a, FA, b, funcion1(b), p, FP])

        if FP == 0:
            return p, datos

        i = i + 1

        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

        error_relativo = abs(b - a) / min(abs(a), abs(b))
        if error_relativo < tol:
            return p, datos

    return p, datos
```

## Aplicación al ejemplo 1

Tomamos el intervalo que aparece en Burden, $[1,2]$. La tolerancia será $10^{-4}$.

```python
a0 = 1
b0 = 2
tol = 10**(-4)
nmax = 100

raiz, datos = biseccion_ejemplo1(a0, b0, tol, nmax)

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.10f}".format(a0, b0, raiz))
print("Iteraciones:", len(datos))
```

## Resultados de las iteraciones

Se muestran las cantidades principales de cada iteración, siguiendo la tabla 2.1 de Burden.

```python
print("# iter\t\ta\t\t f(a)\t\t b\t\t f(b)\t\t p\t\t f(p)\t\t error")
for fila in datos:
    print("{0}\t\t{1:10.7f}\t{2:10.7f}\t{3:10.7f}\t{4:10.7f}\t{5:10.7f}\t{6:10.7f}\t{7:10.7f}".format(*fila))
```

## Gráfica

Aquí se puede observar cómo se van acercando los puntos medios.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

PLT.plot(iteraciones, puntos, 'o-')
PLT.xlabel('Iteración')
PLT.ylabel('p')
PLT.title('Convergencia del método de bisección')
PLT.grid()
PLT.show()
```

## Comprobación

Evaluamos la función en la aproximación que obtuvimos.

```python
print("p =", raiz)
print("f(p) =", funcion1(raiz))
```

$$\boxed{p_{13}=1.3651123047}$$

Este valor coincide con la aproximación presentada en el libro.
