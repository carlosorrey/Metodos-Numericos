# Método de Bisección

### Ejemplo 1 - Burden

## Problema

En el ejemplo 1 del libro se quiere encontrar una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Comprobamos primero los extremos:

$$f(1)=-5$$
$$f(2)=14$$

Como tienen signos diferentes, podemos usar bisección.

## Método

La bisección divide el intervalo en dos y toma el punto medio:

$$m=a+\frac{b-a}{2}$$

Se repite el proceso hasta encontrar una aproximación suficientemente cercana a la raíz.

## Librerías

Se utilizan las mismas librerías empleadas en el código visto en clase: `math`, `numpy` y `matplotlib`.

```python
from math import *
import numpy as np
import matplotlib.pyplot as PLT
```

## Función

```python
def funcion1(x):
    return x**3 + 4*x**2 - 10
```

## Código de bisección

Se usa el mismo método visto en clase, pero con la función y el intervalo del ejemplo de Burden.

```python
def biseccion(a0, b0, tol, nmax):
    # valores iniciales
    a = a0
    b = b0
    i = 1
    FA = funcion1(a)
    datos = []

    # ciclo iterativo
    while i <= nmax:
        # punto medio
        p = a + (b - a)/2
        FP = funcion1(p)

        datos.append([i, a, FA, b, funcion1(b), p, FP, (b-a)/2])

        # criterio de parada
        if FP == 0 or (b - a)/2 < tol:
            return p, datos

        i = i + 1

        # se conserva el intervalo correcto
        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

    # no se alcanzó la tolerancia
    return p, datos
```

### Prueba del ejemplo 1

Para este ejemplo se usa el criterio de error relativo de Burden:

$$\frac{|p-p_n|}{\min\{|a_n|,|b_n|\}}<10^{-4}.$$

```python
def biseccion_ejemplo1(a0, b0, tol, nmax):
    a = a0
    b = b0
    i = 1
    FA = funcion1(a)
    datos = []

    while i <= nmax:
        p = a + (b - a)/2
        FP = funcion1(p)

        datos.append([i, a, FA, b, funcion1(b), p, FP])

        if FP == 0:
            return p, datos

        i = i + 1

        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p

        error_relativo = abs(b - a) / min(abs(a), abs(b))
        if error_relativo < tol:
            return p, datos

    return p, datos
```

## Aplicación al ejemplo 1

Tomamos el intervalo que aparece en Burden, $[1,2]$. La tolerancia será $10^{-4}$.

```python
a0 = 1
b0 = 2
tol = 10**(-4)
nmax = 100

raiz, datos = biseccion_ejemplo1(a0, b0, tol, nmax)

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.10f}".format(a0, b0, raiz))
print("Iteraciones:", len(datos))
```

## Resultados de las iteraciones

Se muestran las cantidades principales de cada iteración, siguiendo la tabla 2.1 de Burden.

```python
print("# iter\t\ta\t\t f(a)\t\t b\t\t f(b)\t\t p\t\t f(p)\t\t error")
for fila in datos:
    print("{0}\t\t{1:10.7f}\t{2:10.7f}\t{3:10.7f}\t{4:10.7f}\t{5:10.7f}\t{6:10.7f}\t{7:10.7f}".format(*fila))
```

## Gráfica

Aquí se puede observar cómo se van acercando los puntos medios.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

PLT.plot(iteraciones, puntos, 'o-')
PLT.xlabel('Iteración')
PLT.ylabel('p')
PLT.title('Convergencia del método de bisección')
PLT.grid()
PLT.show()
```

## Comprobación

Evaluamos la función en la aproximación que obtuvimos.

```python
print("p =", raiz)
print("f(p) =", funcion1(raiz))
```

## Resultado final

Al aplicar el método al intervalo $[1,2]$ y utilizar el criterio de error relativo del Ejemplo 1 de Burden, se obtiene después de 13 iteraciones:

$$\boxed{p_{13}=1.3651123047}$$

Este valor coincide con la aproximación presentada en el libro.
