# Método de Newton

Se busca resolver la ecuación:

$$f(x)=\cos(x)-x=0$$

utilizando el Método de Newton descrito en el Algoritmo 2.3 de Burden.

## Algoritmo 2.3 (Burden)

La fórmula iterativa es:

$$p_n=p_{n-1}-\frac{f(p_{n-1})}{f'(p_{n-1})}$$

El proceso termina cuando:

$$|p_n-p_{n-1}|<TOL$$

## Formulación del problema

$$f(x)=\cos(x)-x$$

$$f'(x)=-\sin(x)-1$$

Se utiliza la aproximación inicial \(p_0=\pi/4\).

```python
from math import cos, sin, pi
import matplotlib.pyplot as plt
```

```python
def f(x):
    return cos(x)-x

def derivada(x):
    return -sin(x)-1
```

```python
def newton(p0,tol,nmax):
    datos=[]
    for i in range(1,nmax+1):
        p=p0-f(p0)/derivada(p0)
        error=abs(p-p0)
        datos.append((i,p,f(p),error))
        if error<tol:
            return p,datos,True
        p0=p
    return p,datos,False
```

```python
p0=pi/4
tol=1e-10
nmax=100

raiz,datos,exito=newton(p0,tol,nmax)

print("Iteración    p_n")
for i,p,_,_ in datos:
    print(i, round(p,10))

print("\nRaíz aproximada:",round(raiz,10))
print("f(raíz)=",f(raiz))
```

## Comparación con Burden

```python
x=[i*(pi/2)/300 for i in range(301)]
y=[f(v) for v in x]

plt.figure(figsize=(8,5))
plt.plot(x,y,label="f(x)=cos(x)-x",color="blue")
plt.axhline(0,color="black")
plt.plot(raiz,f(raiz),"ro",label="Raíz")
plt.title("Método de Newton")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()
```
