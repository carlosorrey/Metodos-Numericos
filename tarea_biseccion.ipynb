# Método de bisección

## Ejemplo 1

Buscamos una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Como $f(1)=-5$ y $f(2)=14$, los valores tienen signos diferentes y podemos usar el método de bisección.

El método consiste en calcular el punto medio del intervalo y conservar la mitad en la que sigue existiendo el cambio de signo.

$$m=a+\frac{b-a}{2}$$

```python
# Metodo de bisección

from math import *
import numpy as np
import matplotlib.pyplot as PLT
```

```python
def función1(x):
   return x**3 + 4*x**2 - 10
```

```python
# ingreso de datos de entrada para el método
a = 1
b = 2

# guarda valores iniciales
a0 = a
b0 = b

tol = 0.0001
nmax = 100
error = 100
niter = 0
```

```python
# Metodo de bisección

# evalua primer valor medio
m = a + (b - a)/2

# Evaluacion de la función en los puntos a, b y m
fa = función1(a)
fb = función1(b)
fm = función1(m)

print("# iter\t\ta \t\tf(a) \t\tb \t\tf(b) \t\tm \t\tf(m) \t\t error")
print("{0} \t\t{1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error))

# ciclo iterativo
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = función1(a)
    else:
        b = m
        fb = función1(b)

    m = a + (b - a)/2
    fm = función1(m)
    error = abs(b - a)
    niter += 1
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```

## Gráfica

Para ver cómo se acerca el punto medio a la raíz, usamos los valores de $m$ de las iteraciones.

```python
# grafica de la función y la raíz aproximada
x = np.linspace(1, 2, 200)
y = función1(x)

PLT.figure(figsize=(7, 4))
PLT.plot(x, y, label='f(x)')
PLT.axhline(0, linestyle='--')
PLT.axvline(m, linestyle='--', label=f'Raíz ≈ {m:.4f}')
PLT.xlabel('x')
PLT.ylabel('f(x)')
PLT.title('Método de bisección')
PLT.grid(True)
PLT.legend()
PLT.show()
```

## Comprobación

Evaluamos la función en la aproximación obtenida.

```python
print(f"m = {m:.6f}")
print(f"f(m) = {función1(m):.6f}")
```

La aproximación obtenida es $m\approx 1.365173$. Al evaluar la función, el resultado queda muy cerca de cero.
