# Método de bisección

## Ejemplo 1

El ejercicio consiste en encontrar una aproximación de la raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Primero revisamos los extremos:

$$f(1)=-5,\qquad f(2)=14$$

Como los signos son diferentes, se puede aplicar el método de bisección.

### Método

La idea es tomar el punto medio del intervalo y decidir con cuál de los dos extremos se queda la raíz. Se repite hasta que el intervalo sea suficientemente pequeño.

El punto medio se calcula con

$$m=a+\frac{b-a}{2}$$

```python
# Librerías que usamos en el programa
from math import *
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Función del ejemplo de Burden
def f(x):
    return x**3 + 4*x**2 - 10
```

## Datos de entrada

Aquí se colocan los valores iniciales, la tolerancia y el número máximo de iteraciones. La estructura es la misma que usamos en clase.

```python
# Datos de entrada
a = 1
b = 2

# Guardamos los valores iniciales
a0 = a
b0 = b

tol = 0.0001
nmax = 100
error = 100
niter = 0
```

## Método de bisección

Primero se calculan los valores de la función en los extremos y en el primer punto medio. Después comienza el ciclo iterativo.

```python
# Evaluamos la función en los extremos
fa = f(a)
fb = f(b)

# Primer punto medio
m = a + (b - a) / 2
fm = f(m)

datos = []

print("# iter\t\ta\t\tf(a)\t\tb\t\tf(b)\t\tm\t\tf(m)\t\terror")
print("{0}\t\t{1:6.4f}\t{2:8.4f}\t{3:6.4f}\t{4:8.4f}\t{5:6.4f}\t{6:8.4f}\t{7:8.4f}".format(niter, a, fa, b, fb, m, fm, error))

# Ciclo iterativo
while error > tol and niter < nmax:
    # Se calcula el punto medio una sola vez en cada vuelta
    m = a + (b - a) / 2
    fm = f(m)

    if np.sign(fa) == np.sign(fm):
        a = m
        fa = fm
    else:
        b = m
        fb = fm

    error = abs(b - a)
    niter += 1

    datos.append([niter, a, fa, b, fb, m, fm, error])

    print("{0}\t\t{1:6.4f}\t{2:8.4f}\t{3:6.4f}\t{4:8.4f}\t{5:6.4f}\t{6:8.4f}\t{7:8.4f}".format(niter, a, fa, b, fb, m, fm, error))
```

## Resultado

Al terminar las iteraciones, tomamos el último punto medio como aproximación de la raíz.

```python
print(f"Raíz hallada: {m:.6f}")
print("Iteraciones necesarias:", niter)
```

## Tabla de resultados

La tabla permite revisar los valores obtenidos en cada iteración.

```python
columnas = ['i', 'a', 'f(a)', 'b', 'f(b)', 'm', 'f(m)', 'error']
pd.DataFrame(datos, columns=columnas)
```

## Gráfica

Se muestran los puntos medios de las iteraciones para ver cómo se van acercando a la raíz.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

plt.figure(figsize=(7, 4))
plt.plot(iteraciones, puntos, 'o-', label='Punto medio (m)')
plt.axhline(m, linestyle='--', label=f'Raíz aproximada ≈ {m:.4f}')
plt.xlabel('Iteración')
plt.ylabel('m')
plt.title('Convergencia del método de bisección')
plt.grid(True)
plt.legend()
plt.show()
```

## Comprobación

Evaluamos la función en el último punto medio para ver qué tan cerca queda de cero.

```python
print(f"m = {m:.6f}")
print(f"f(m) = {f(m):.6f}")
```

La aproximación obtenida es $m\approx 1.365173$. Al evaluar la función, el resultado queda muy cerca de cero.

En el ejemplo de Burden aparece aproximadamente $1.36511$. La diferencia se debe al criterio de parada: en el ejemplo del libro se usa un criterio de error relativo, mientras que en este programa usamos $|b-a|$, que es el tamaño del intervalo. El procedimiento de bisección es el mismo.
