# Método de bisección

### Ejercicio 1

En este cuaderno se aplica el método de bisección del Algoritmo 2.1 al ejemplo 1 del libro.

## Problema

La función que vamos a trabajar es

$$f(x)=x^3+4x^2-10$$

y el intervalo inicial es $[1,2]$.

Comprobamos los extremos:

$$f(1)=-5,\qquad f(2)=14$$

Como tienen signos diferentes, podemos aplicar el método de bisección.

## Método

La idea es ir dividiendo el intervalo en dos partes y quedarnos con la parte donde todavía hay un cambio de signo. Para eso usamos el punto medio

$$m=a+\frac{b-a}{2}$$

En cada vuelta se calcula el valor de la función en ese punto y se actualiza uno de los extremos. En este ejercicio usamos una tolerancia de $10^{-4}$.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

Definimos la función del ejemplo:

```python
def f(x):
    return x**3 + 4*x**2 - 10
```

## Código de bisección

La estructura es parecida a la que usamos en clase. `fa`, `fb` y `fm` guardan los valores de la función para no tener que calcularlos otra vez cuando ya los tenemos.

```python
def biseccion(a0, b0, tol, nmax):
    a = a0
    b = b0
    niter = 1

    fa = f(a)
    fb = f(b)
    error = abs(b - a)
    datos = []

    while error > tol and niter < nmax:
        # Calculamos el punto medio una sola vez por iteración
        m = a + (b - a) / 2
        fm = f(m)

        if np.sign(fa) == np.sign(fm):
            a = m
            fa = fm
        else:
            b = m
            fb = fm

        error = abs(b - a)
        datos.append([niter, a, fa, b, fb, m, fm, error])
        niter += 1

    return m, datos
```

## Aplicación al ejemplo 1

Usamos el intervalo $[1,2]$, una tolerancia de $10^{-4}$ y un máximo de 100 iteraciones.

```python
a0 = 1
b0 = 2
tol = 10**(-4)
nmax = 100

raiz, datos = biseccion(a0, b0, tol, nmax)

print(f"La raíz aproximada es: {raiz:.6f}")
print("Iteraciones:", len(datos))
```

## Resultados

Aquí podemos ver cómo van cambiando los extremos y el punto medio en cada iteración.

```python
tabla = pd.DataFrame(datos, columns=['iter', 'a', 'f(a)', 'b', 'f(b)', 'm', 'f(m)', 'error'])

tabla
```

## Gráfica

En la gráfica se muestran los valores de $m$ obtenidos en las iteraciones. También se marca la raíz aproximada para comparar hacia dónde se van acercando los puntos.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

plt.figure(figsize=(8, 4.5))
plt.plot(iteraciones, puntos, 'o-', label='Puntos medios')
plt.axhline(raiz, linestyle='--', label=f'Raíz aproximada = {raiz:.10f}')
plt.xlabel('Iteración')
plt.ylabel('m')
plt.title('Convergencia del método de bisección')
plt.grid(True)
plt.legend()
plt.show()
```

## Comprobación y resultado

Comprobamos qué tan cerca de cero queda la función evaluada en nuestro resultado.

```python
print(f"m = {raiz:.6f}")
print(f"f(m) = {f(raiz):.6f}")
```
