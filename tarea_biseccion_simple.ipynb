# Método de Bisección - Ejemplo 1

Buscamos una raíz de

$$f(x)=x^3+4x^2-10$$

en el intervalo $[1,2]$.

Primero revisamos el cambio de signo:

$$f(1)=-5,\qquad f(2)=14$$

Como $f(1)$ y $f(2)$ tienen signos distintos, podemos aplicar bisección.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def f(x):
    return x**3 + 4*x**2 - 10
```

## Código del algoritmo

La función sigue la misma idea del código que vimos en clase: se calcula el punto medio, se revisa el signo y se cambia uno de los extremos del intervalo.

```python
def biseccion(a0, b0, tol, nmax):
    a, b = a0, b0
    fa, fb = f(a), f(b)
    niter = 1
    datos = []
    error = abs(b - a)
    
    while error > tol and niter <= nmax:
        m = a + (b - a) / 2.0
        fm = f(m)
        
        datos.append([niter, a, fa, b, fb, m, fm, error])
        
        if np.sign(fa) == np.sign(fm):
            a, fa = m, fm
        else:
            b, fb = m, fm
            
        error = abs(b - a)
        niter += 1
        
    return m, datos
```

## Pruebas y resultados

Para el ejemplo usamos $a=1$, $b=2$, una tolerancia de $10^{-4}$ y como máximo 100 iteraciones.

```python
raiz, datos = biseccion(a0=1, b0=2, tol=1e-4, nmax=100)

print(f"Raíz hallada: {raiz:.6f}")
print("Iteraciones necesarias:", len(datos))
```

```python
# Ver los datos de las iteraciones
columnas = ['i', 'a', 'f(a)', 'b', 'f(b)', 'm', 'f(m)', 'error']
pd.DataFrame(datos, columns=columnas)
```

## Gráfica

Aquí se puede ver cómo los puntos medios se van acercando al valor de la raíz.

```python
iteraciones = [fila[0] for fila in datos]
puntos = [fila[5] for fila in datos]

plt.figure(figsize=(7, 4))
plt.plot(iteraciones, puntos, 'o-', label='Punto medio (m)')
plt.axhline(raiz, linestyle='--', label=f'Raíz ≈ {raiz:.4f}')
plt.xlabel('Iteración')
plt.ylabel('m')
plt.title('Convergencia del método')
plt.grid(True)
plt.legend()
plt.show()
```

## Comprobación

Comprobamos qué tan cerca de cero queda la función usando el resultado que obtuvimos.

```python
print(f"m = {raiz:.6f}")
print(f"f(m) = {f(raiz):.6f}")
```

### Resultado

La aproximación que obtenemos es $m\approx 1.365173$. Al evaluar la función, el resultado queda muy cerca de cero, así que la aproximación es adecuada para la tolerancia que usamos.

En el libro de Burden aparece aproximadamente $1.36511$. La diferencia se debe a que en el ejemplo del libro se utiliza un criterio de error relativo, mientras que aquí usamos el tamaño del intervalo, $|b-a|$, como criterio de parada. El procedimiento de bisección sigue siendo el mismo.
